# Matplotlib: From Numerical Data to Visualization

Matplotlib is a Python library for creating visualizations.

In the upcoming assignment, three libraries have different jobs:

**pandas**
-> loads and organizes the data

**NumPy**
-> performs the numerical calculations

**Matplotlib**
-> displays the results

The important idea is that Matplotlib does not calculate the linear regression.

NumPy will calculate the regression and produce the numerical values. Matplotlib will receive those values and display them.

## Scatter Plot vs. Line Plot

This distinction will become important in the CO2 assignment.

The measured CO2 values are observations.

Those observations can be displayed as individual points:

x, y -> observations

The regression calculated by NumPy produces another set of values:

x, predicted -> regression line

This means a single figure can contain both:

* observed data shown as points
* calculated regression shown as a line

## Figures and Axes

You may have noticed examples using code similar to:

`fig, ax = plt.subplots()`

A **Figure** is the complete figure or window.

An **Axes** is the plotting area inside the figure.

The plotting commands can then be called through the Axes object.

For example:

`ax.plot(...)`

`ax.scatter(...)`

`ax.set_title(...)`

Don't confuse an Axes object with the x-axis or y-axis. The Axes is the complete plotting area that contains both axes.

```python
fig, ax = plt.subplots()

ax.set_title("Practice Plot")
ax.set_xlabel("X Values")
ax.set_ylabel("Y Values")

plt.show()
```

# What Matplotlib Is NOT Doing

This distinction is important.

Matplotlib is not calculating:

* the mean of x
* the mean of y
* the regression numerator
* the regression denominator
* the slope
* the intercept
* the predicted values

Those calculations belong to NumPy.

Matplotlib receives the resulting arrays and visualizes them.

This separation of responsibilities will also be reflected in the class design of the assignment.
